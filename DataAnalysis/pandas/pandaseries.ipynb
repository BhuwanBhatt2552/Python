# the Index
-> the *index* let you easily access **"rows"** in a pandas series or data frame.

-> you can *index and slice* series like other sequence datatypes, but we'll learn a better method.

**Custom Indices:-** There are cases where it's applicable to use a custom index for accessing rows.
* Custom indices can be assigned when **creating the series** or **by assignment**

```python
my_series = pd.Series(range(5))

my_series
```

```python
my_series[3]
```

```python
my_series[1::2]
```

```python
my_series = pd.Series(range(5), index = ["Day 0", "Day 1", "Day 2", "Day 3", "Day 4" ])

my_series
```

```python
my_series["Day 2"]
```

```python
my_series[::2]
```

# The .iloc[] method

-> The .iloc[] method is the prefered way to access values by their positional index.
* This method works even when Series have a custom, non-integer index
* It is more efficient than slicing and is recommended by Pandas' creators.

-> Syntax: df.iloc[row position, column position]

```python
my_series.iloc[-1]
```

```python
my_series.iloc[1:4]
```

```python
my_series = pd.Series(range(5), index = ["Day 0", "Day 1", "Day 2", "Day 3", "Day 4" ])

my_series
```

```python
my_series.loc["Day 2"]
```

```python
my_series.loc["Day 1" : "Day 3"]
```

```python
my_series.index = [0, 2, 3, 100, 5]

my_series
```

```python
my_series.loc[0: 5]
```

```python
my_series.reset_index(drop=True).loc[:3]
```
