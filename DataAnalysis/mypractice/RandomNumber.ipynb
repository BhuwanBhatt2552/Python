```python
import numpy as np
```

-> Create a random number generator with a seed of 100


```python
rng = np.random.default_rng(616)  # Create a random number generator with a seed of 100
rng.random(100)
```

-> Generate 100 random integers between 0 and 10 (exclusive of 10)


```python
rng.integers(0, 10, 100)  #  Generate 100 random integers between 0 and 10 (exclusive of 10)
```

-> Generate 10 random numbers from a normal distribution with a mean of 50 and a standard deviation of 5


```python
rng.normal(50, 5, 10) # Generate 10 random numbers from a normal distribution with a mean of 50 and a standard deviation of 5
```

# Indexing and Slicing


```python
integer_arr = np.arange(20)
integer_arr
```

Indexing and slicing in one dimensional array is same as base python.

* array[index] - indexing to access single element(0-indexed)
* array[start:stop:step_size]- slicing to access series of elements(stop is not inclusive)


```python
integer_arr[11]
```

```python
integer_arr[-1]
```

```python
integer_arr[2:8:2]
```

```python
integer_arr[:5]
```

```python
integer_arr[:11:3]
```

```python
integer_arr[1: :2]
```

```python
new_arr = integer_arr.reshape(4, 5)

new_arr
```

```python
new_arr = integer_arr.reshape(10,2)
new_arr
```

```python
new_arr[:,:]
```

```python
new_arr = integer_arr.reshape(4, 5)

new_arr

new_arr[1:,:]
```

Indexing and slicing two dimensional arrays requires an extra index or slice.

- array[row index, column index]- indexing to access a single element (0- indexed)
- array[start:stop:stepsize, start:stop:stepsize]- slicing to access a series of elements
