# ARRAY AGGREGATION METHOD

**Array aggregation methods** let you to calculate metrics like sum, mean, and max

1. **array.sum()**- *returns the sum of all values in an array*
2. **array.max()**- *returns the largest value in an array*
3. **array.min**- *returns the smallest value in an array*
4. **array.mean()**- *returns the average of values in an array*

-> You can also aggregate across **rows** and **columns**.

* array.sum(axis = 0) - aggregates across rows
* array.sum(axis = 1) - aggregates across columns

```python
import numpy as np
```

```python
rng = np.random.default_rng(2591)

price = (rng.random(10) * 10).round(2)

price
```

```python
inventory = rng.integers(0, 100, 10)
inventory
```

```python
price.mean()
```

```python
inventory.mean()
```

```python
inventory.sum()
```

```python
price.max()
```

```python
price.min()
```

```python
inventory.std()
```

```python
(price * inventory)
```

```python
(price * inventory).sum()
```

```python
(price * inventory).mean()
```

```python
(price * inventory).argmax()
```

```python
(price * inventory).argmin()
```

```python
price_2d = price.reshape(5,2)

price_2d
```

```python
price_2d.sum(axis=0)
```

```python
price_2d.sum(axis=1)
```

```python
price_2d.mean(axis=1)
```
