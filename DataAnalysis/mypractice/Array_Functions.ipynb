# Array Functions

-> Arrays functions allow you to perform other aggregations like median and percentiles. 


* **np.median(array)** - returns median value in an array.

* **np.percentile(array, n)** - returns a value in the n<sup>th</sup> percentile in an array.  This uses linear interpolation by default.

* **np.unique(array)** - returns unique values in array.

* **np.sqrt(array)** - returns the square root for each value in  an array.

```python
import numpy as np
```

```python
rng  = np.random.default_rng(21)

prices = (rng.random(10) * 10).round(2)



inventory = rng.integers(10, 100, 10)

print("prices= ",prices)
print("inventory= ",inventory)
```

```python
product_Value = prices * inventory

print("product_values ", product_Value)
```

```python
product_Value.sort()

product_Value
```

```python
reshaped_prod_val = product_Value.reshape(2, 5)

reshaped_prod_val
```

```python
np.median(reshaped_prod_val).round(2)
```

```python
np.percentile(reshaped_prod_val, 90)
```

```python
np.sqrt(reshaped_prod_val)
```

```python
np.unique(reshaped_prod_val)
```
