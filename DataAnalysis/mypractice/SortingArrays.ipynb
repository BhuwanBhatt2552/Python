# *Sorting Arrays*

-> The `sort()` method will sort arrays in place
  * use the axis argument to specify the dimension to sort by

-> axis =1 by default, sorts a 2D array row by row

```python
import numpy as np
```

```python
product_value = np.array([205.2, 123.6, 183.3, 63.21, 23.21, 9.12, 7341.1, 303.2, 654.23])

product_value
```

```python
np.sort(product_value)
```

```python
product_value
```

```python
product_value.sort()
```

```python
product_value
```

```python
product_value[0]
```

```python
product_value[-1]
```

```python
product_value[::-1]
```
