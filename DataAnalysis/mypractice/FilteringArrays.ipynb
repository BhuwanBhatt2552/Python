# Filtering Arrays


-> You can filter Arrays by indexing them with logical tests.

* Only the array elements is positions where the logical test returns **True** are returned.
* Performing a logical test on a *NumPy array* returns a *Boolean array* with the results of the *logical test* on each array elements.
* Indexing an array with a Boolean array returns an array with the elements where the Boolean value is *True*.

-> You can Filter arrays with **multiple logical tests**
* Use **|** for **or** conditions and **&** for **and** conditions  

**PRO TIP:-**  
  *Store complex filtring citeria in a **variable** (known as a **Boolean Mask**)*

-> You can filter arrays based on **Values in other arrays**
* Use the Boolean array returned from the other array to index the array you want to filter.

# Modifying Array Values

-> You can **modify array values** by assigning new ones.

```python
import numpy as np
```

```python
my_array = np.arange(20)

my_array
```

```python
my_array % 2 == 0 
```

```python
my_array[my_array % 2 == 0 ]
```

```python
even_odd = np.array(['even', 'odd'] * 10 )
even_odd
```

```python
even_odd != 'odd'
```

```python
my_array[even_odd != 'odd'] = 0

my_array
```

```python
my_array[2] = 2
my_array
```

```python
mask = (even_odd != 'odd') | (even_odd != 'even')

even_odd[mask]
```

```python
mask = (even_odd != 'odd') & (even_odd != 'even')

even_odd[mask]
```

# The **WHERE** Function
* The *where()* Numpy function performs a logical test and returns a given value if the is *True*, or another if the test is *Fales*. 

Syntax:- np.where(logical test, value if True, value if False)

```python
np.where(my_array %2 == 0, 'even', 'odd')
```

```python
np.where(my_array % 2 == 0, 'even', np.where(my_array == 9 , my_array, 'odd'))
```
