# Array Operations
- Arithmetic operators can be used to perform the array operations.
- Array operations are applied via *vectorization* and *broadcasting*, which eliminates the need of loop through the array's elements.

```python
import numpy as np

rng = np.random.default_rng(616)
inventory =  rng.integers(0, 100, 10)
inventory
```

```python
inventory - 24
```

```python
inventory *2
```

```python
inventory + 11
```

```python
inventory / 3 
```

```python
(inventory/3).round(2)
```

```python
inventory.sum()
```

```python
(inventory / 3 ).dtype
```

```python
price = (rng.random(10) * 10).round(3)

price
```

```python
price * inventory
```

```python
(price * inventory).sum()
```

```python
inventory_list = list(inventory)
```

```python
inventory_list
```

```python
new_inventory = [ ]

for x in inventory_list:
    new_inventory.append(x+2)

new_inventory    
```

```python
new_inventory = [int(x + 2) for x in inventory_list]
print(new_inventory)
```

```python
[x+2 for x in inventory_list]
```

```python
[x+y for x, y in zip(inventory_list, price)]
```
