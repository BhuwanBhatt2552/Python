The creativity of our marketing team knows no bounds!

They have asked us to come up with a simple algorithm to provide a random discount to our list of prices below.

Before we do that, we want to add a $5 shipping fee that will be calculated before the discount is applied. Call this array `total`.

Once we have that, we want to use the random array and apply it to the six prices.
1. Grab the first six numbers from the `random_array`, reshape it to one dimension.
   Call this `discount_pct`.
2. Subtract `discount_pct` from 1, store this in `pct_owed`.
3. Multiply `pct_owed` by `total` to get the final amount owed.  </span>

```python
import numpy as np
```

```python
prices  = np.array([12.99, 21.99, 34.59, 5.99, 99.99, 46.29])

total = prices + 5

total
```

```python
rng = np.random.default_rng(2022)
random_array = rng.random((3, 3))

random_array.round(2)
```

```python
discount_pct = random_array[:2, : ].reshape(6)

discount_pct.round(2)

pct_owed = 1 - discount_pct

pct_owed.round(2)

final_owed = total * pct_owed

final_owed.round(2)
```

```python
((1 - (random_array[:2, :].reshape(6))) * total).round(2)
```
