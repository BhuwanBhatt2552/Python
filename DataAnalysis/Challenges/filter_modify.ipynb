```python
import numpy as np
```

```python
prices  = np.array([12.99, 21.99, 34.59, 5.99, 99.99, 46.29])

prices
```

```python
products = np.array(["Salad", "Bread", "Tomato", "Cola", "Mustard", "Gourment Ice Cream"])

products
```

```python
products[prices > 25]
```

```python
mask = (prices > 25 ) | (products == 'Cola')
fancy_feast_special = products(mask)

fancy_feast_special
```
